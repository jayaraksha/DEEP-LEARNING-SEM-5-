```python
import tensorflow as tf
from tensorflow.keras import layers,models

(xtr,ytr),(xte,yte)=tf.keras.datasets.mnist.load_data()
xtr,xte=xtr/255.0,xte/255.0
xtr=xtr[...,None]
xte=xte[...,None]

m=models.Sequential([
layers.Conv2D(32,(3,3),activation='relu',input_shape=(28,28,1)),
layers.MaxPooling2D(2,2),
layers.Flatten(),
layers.Dense(128,activation='relu'),
layers.Dense(10,activation='softmax')])

m.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
m.fit(xtr,ytr,epochs=5,validation_split=.1)
print("Test Accuracy:",m.evaluate(xte,yte,verbose=0)[1])
print("Predicted Digit:",m.predict(xte[:1]).argmax())
```

```python
import tensorflow as tf
from tensorflow.keras import layers,models

(xtr,ytr),(xte,yte)=tf.keras.datasets.mnist.load_data()
xtr,xte=xtr/255.0,xte/255.0
xtr=xtr[...,None]
xte=xte[...,None]

m=models.Sequential([
layers.Conv2D(32,(3,3),activation='relu',input_shape=(28,28,1)),
layers.MaxPooling2D(2,2),
layers.Flatten(),
layers.Dense(128,activation='relu'),
layers.Dense(10,activation='softmax')])

m.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
m.fit(xtr,ytr,epochs=5,validation_split=.1)
print("Test Accuracy:",m.evaluate(xte,yte,verbose=0)[1])
print("Predicted Digit:",m.predict(xte[:1]).argmax())
m.summary()

print("Input Dimensions:",xte.shape)

print("Predicted Labels:",m.predict(xte[:10]).argmax(axis=1))

print("Test Cases:")
for i in range(5):
    print("Actual:",yte[i],"Predicted:",m.predict(xte[i:i+1]).argmax())

test_loss,test_acc=m.evaluate(xte,yte,verbose=0)
print("Test Accuracy:",test_acc)
```
